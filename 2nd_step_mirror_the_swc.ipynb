```python
# 去掉标记的点
DataPath = r"R:\gts\data center file_0221\swc_new\*"
outPath = r"R:\gts\data center file_0907\swc"

files = glob.glob(DataPath)
for filename in files:
    neurons = filename.split("\\")[-1]
    output = os.path.join(outPath,neurons)
    # print(output)
    f = open(filename,"r")
    o = open(output,"w")
    try:
        for lines in f:
            records = lines.split("\t")[0].split(" ")
            if int(records[1])>3:
                records[1] = str(0)
            elif int(records[1])==0:
                records[1] = str(2)
            else:
                records[1] = records[1]
            o.write(records[0]+' '+records[1]+' '+records[2]+' '+records[3]+' '+records[4]+' '+records[5]+' '+records[6])
        o.close()
        f.close()
    except:
        print(output)
```

```python
# dowmsample neurons
import numpy as np
import navis
import os,glob
```

```python
all_data = r"E:\all neurons 0102\HY_swc\All_HY_swc"
out_path = r"E:\all neurons 0102\HY_swc\resample_swc"
```

```python
for file in os.listdir(all_data):
    print(file)
    n = navis.read_swc(os.path.join(all_data,file), units='micrometer')
    resampled_n = navis.downsample_neuron(n, downsampling_factor = 20, inplace = True)
    navis.write_swc(resampled_n, os.path.join(out_path,file), header=None)
```

```python
for file in os.listdir(out_path):
    print(file)
    f = open(os.path.join(out_path,file), 'r')
    swc = f.readlines()
    f.close()
    
    f = open(os.path.join(out_path,file), 'w')
    for item in swc[6:]:
        f.write(item)
    f.close()
```
