```python
import pandas as pd
import os,glob
import seaborn as sns
import matplotlib.pyplot as plt 
import scipy.io
```

```python
import matplotlib as mpl
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
plt.rc("font",family="Arial")
```

```python
os.chdir(r"G:\latency\data\new")
```

```python
df = pd.DataFrame()
```

```python
files = os.listdir()
```

```python
for file in files:
    dftmp = pd.read_excel(file,index_col=0)
    dftmp["geno"] = file.split("_")[1]
    df = pd.concat([df,dftmp],axis=0)
```
