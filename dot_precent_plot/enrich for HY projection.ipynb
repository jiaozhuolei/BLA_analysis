```python
# location base
def get_wendu_type(x):
    return x.p*len(df)/(x.name+1)

dfall = pd.DataFrame()
dftmp = pd.DataFrame()
for i in set(df200.cluster.to_list()):
    dfin = df200.loc[df200.cluster ==i]
    dfout = df200.loc[df200.cluster !=i]
    sumab = len(dfin)
    sumcd = len(dfout)
    dict_tmp = {}
    for area in df200.columns.tolist():
        if area != "cluster":
            a = (dfin[area]>0).astype("int").sum()
            b = sumab - a
            c = (dfout[area]>0).astype("int").sum()
            d = sumcd -c
            obs2 = [[a,b], [c,d]]
            ratio, pvalue =stats.fisher_exact(obs2)
            if ratio >1:
                dict_tmp[area] = pvalue
        dic_total = {}
        area_tmp = "cluster_%s"%(i)
        dic_total[area_tmp] = dict_tmp
        df = pd.DataFrame(dic_total)
        try:
            df.loc[:,"cluster"] = i
            df.columns = ["p","cluster"]
            df = df.sort_values(by = "p",axis=0,ascending=True)
            df = df.reset_index(drop=False)
            df.columns = ["area","p","cluster"]
            df.loc[:, "FDR"] = df.apply(get_wendu_type, axis=1)
            dftmp = df.iloc[0:30,:]
            # dftmp = df.iloc[0:20,:]
        except:
            pass
    dfall = pd.concat([dfall,dftmp],axis=0)              
```
