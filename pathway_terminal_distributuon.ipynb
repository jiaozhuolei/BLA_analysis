```python
import matplotlib.pyplot as plt
import pandas as pd
from collections import Counter

data = [
    "ORB", "ORB", "ORB", "ORB", "ORB", "ORB", "ORB", 
    "ACA", "ILA", "ORB", "ORB", "ACA", "ORB", "ORB", 
    "ORB", "ORB", "ORB"
]

counts = Counter(data)
df = pd.DataFrame(counts.items(), columns=["Brain_Region", "Count"])
df["Percentage"] = df["Count"] / df["Count"].sum() * 100
df = df.sort_values(by="Count", ascending=False).reset_index(drop=True)

print(df)
print(f"Total: {len(data)}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

import matplotlib as mpl
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
plt.rc("font",family="Arial")

labels = ['ORB', 'ACA', 'ILA']
counts = [14, 2, 1]
colors = ['#4E79A7', '#F28E2B', '#E15759']  # Elegant categorical palette

fig, ax = plt.subplots(figsize=(6, 6), dpi=300)

# Create donut pie chart
wedges, texts, autotexts = ax.pie(
    counts,
    labels=labels,
    autopct=lambda pct: f'{pct:.1f}%\n(n={int(round(pct*sum(counts)/100))})',
    startangle=90,
    pctdistance=0.75,
    colors=colors,
    wedgeprops=dict(width=0.4, edgecolor='white', linewidth=2),
    textprops=dict(fontsize=12, weight='medium')
)

for autotext in autotexts:
    autotext.set_fontsize(10)
    autotext.set_weight('bold')

# Center text
ax.text(0, 0, f'Total Pathways\nN = {sum(counts)}', ha='center', va='center', fontsize=12, color='#333333')

ax.set_title('Distribution of Terminal Brain Regions', fontsize=14, pad=20)
plt.tight_layout()
plt.savefig('terminal_regions_donut_chart.png', dpi=300, bbox_inches='tight')
plt.savefig('terminal_regions_donut_chart.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
