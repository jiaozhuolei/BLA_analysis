```python
import os
import pandas as pd

# Check files in current working directory
# print(os.listdir('.'))

# Try to read the csv file
df = pd.read_csv('all_terminal_260831.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

import matplotlib as mpl
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
plt.rc("font",family="Arial")

# Prepare data
region_counts = df['name_use'].value_counts()
labels = region_counts.index.tolist()
counts = region_counts.values.tolist()
total = sum(counts)

# Color palette (distinct, publication quality)
# E.g. Tableau 10 or custom scientific palette
colors = ['#59A14F','#4E79A7', '#F28E2B', '#E15759', '#B07AA1'] 
# Or let's see which colors look best together

fig, ax = plt.subplots(figsize=(6.5, 6.5), dpi=300)

wedges, texts, autotexts = ax.pie(
    counts,
    labels=labels,
    autopct=lambda pct: f'{pct:.1f}%\n(n={int(round(pct*total/100))})',
    startangle=90,
    pctdistance=0.75,
    colors=colors,
    wedgeprops=dict(width=0.4, edgecolor='white', linewidth=2),
    textprops=dict(fontsize=12)
)

for autotext in autotexts:
    autotext.set_fontsize(10)
    autotext.set_weight('bold')

# Center text
ax.text(0, 0, f'Total Terminals\nN = {total}', ha='center', va='center', fontsize=12, color='#333333')

ax.set_title('Distribution of Terminal Brain Regions', fontsize=14, pad=20)
plt.tight_layout()
plt.savefig('all_terminals_donut_chart.png', dpi=300, bbox_inches='tight')
plt.savefig('all_terminals_donut_chart.pdf', dpi=300, bbox_inches='tight')
plt.close()
```
