```python
import numpy as np
import plotly.graph_objects as go

def make_sphere_surface(center, radius=100, resolution=60, color="lightblue", opacity=0.3):
    center = np.array(center, dtype=float).flatten()
    x0, y0, z0 = center[0], center[1], center[2]

    # 提高采样密度（60~80）
    u = np.linspace(0, 2 * np.pi, resolution)
    v = np.linspace(0, np.pi, resolution)

    # 保持 2D 矩阵形状，不要 flatten
    x = x0 + radius * np.outer(np.cos(u), np.sin(v))
    y = y0 + radius * np.outer(np.sin(u), np.sin(v))
    z = z0 + radius * np.outer(np.ones_like(u), np.cos(v))

    # 构造单一颜色的 colorscale
    surfacecolor = np.zeros_like(x)

    return go.Surface(
        x=x, y=y, z=z,
        surfacecolor=surfacecolor,
        colorscale=[[0, color], [1, color]],
        showscale=False,
        opacity=opacity,
        hoverinfo='skip',
        lighting=dict(
            ambient=0.7,
            diffuse=0.8,
            specular=0.5,
            roughness=0.2,
            fresnel=0.2
        )
    )
```
