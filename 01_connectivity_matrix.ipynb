```r
##
NeuronSampleID = matrix(nrow = NumNeuron, ncol = 1)
NeuronIndex = matrix(nrow = NumNeuron, ncol = 1)

NeuronCoord_X = matrix(nrow = NumNeuron, ncol = 1)
NeuronCoord_Y = matrix(nrow = NumNeuron, ncol = 1)
NeuronCoord_Z = matrix(nrow = NumNeuron, ncol = 1)

NeuronHemisphere = matrix(nrow = NumNeuron, ncol = 1)
NeuronRegion = matrix(nrow = NumNeuron, ncol = 1)
```

```r
for(iNeuron in 1:NumNeuron) {
  sampleID = strsplit(Path_SWC_Neurons[iNeuron],"_")[[1]][1]
  neuronID = strsplit(Path_SWC_Neurons[iNeuron],"_")[[1]][2]
  
  NeuronSampleID[iNeuron,1] = sampleID
  NeuronIndex[iNeuron,1] = neuronID
  
  tmpSWC = read.table(Path_SWC_Neurons[iNeuron], 
                      sep = " ", 
                      nrows = 1, # only the first row was read.
                      col.names = c("PointNo","Label","X","Y","Z","W","Parent"))
  
  tmpX = tmpSWC$X[1]
  tmpY = tmpSWC$Y[1]
  tmpZ = tmpSWC$Z[1]
  
  if(tmpSWC$Z[1] > 5700) {
    NeuronHemisphere[iNeuron] = "Right"
    tmpZ = 11400 - tmpZ
  }
  else {
    NeuronHemisphere[iNeuron] = "Left"
  }
  
  NeuronCoord_X[iNeuron] = tmpX #
  NeuronCoord_Y[iNeuron] = tmpY #
  NeuronCoord_Z[iNeuron] = tmpZ #
  
  
  tmpStructureID = allen_annotation[floor(tmpX/allen_resolution)+1, 
                                    floor(tmpY/allen_resolution)+1,
                                    floor(tmpZ/allen_resolution)+1]
  
  if(FALSE == tmpStructureID %in% anno_structure$CurrentID) {
    print(paste0(iNeuron, " : Wrong StructureID of soma: ", Path_SWC_Neurons[iNeuron]))
    NeuronRegion[iNeuron] = "Unknown"
  }
  else {
    NeuronRegion[iNeuron] = anno_structure$Acronym[which(anno_structure$CurrentID == tmpStructureID)]
  }
}
```

```r
# Modular analysis of intra-hypothalamus connectivity
# divide hypothalamus into cubes and quantify the connectivity among these cubes

setwd("./")

NeuronInfo = read.csv(file = "/home/gao/hy/pag_subdomain/output/NeuronInfo_HY_all.txt", sep = " ")
NumNeuron = nrow(NeuronInfo)

# Read NRRD file of hypothalamus ----
# BST
library(nat)
mask = read.nrrd("/home/gao/hy/pag_subdomain/structure_795.nrrd", ReadByteAsRaw = FALSE)

mask[,,58:114] = 0 # ipsi-lateral only (left hemisphere)

IndexCube = which(mask == 1) # localize the cube
NumCube = length(IndexCube)

# for each cube in HY, assign a unqiue ID to each one
anno = array(data = 0, dim = dim(mask))

for(iCube in 1:NumCube) {
  anno[IndexCube[iCube]] = iCube
}

#detach(name = "package:snp_computation", unload = T)
#library(snp)


#' Calculate projections to annotated brain regions
#'
#' @details Annotated brain regions can be any 3D array with
#' an unique index denoting a brain region or domain
#'
#' @param PathOfNeuron Full path of SWC file for each neuron
#' @param Hemisphere Which hemisphere does the neuron reside in
#' @param Annotation 3D array with each element assigned a positive value
#' @param Resolution Voxel size of each annotation, e.g., 100, 50, 25, 10
#' @param NumOfCores Number of cores for parallel computation
#' @param OptionMetric Default value is "LengthOfAxon". ("LengthOfAxon", "NumberOfEndNodes", "NumberOfBranchNodes")
#' @param StructureIDOfTargets Structure ID of brain regions or domains in the annotation.
#' Column name of returned matrix is the same as StuctureIDOfTargets.
#'
#' @return projection matrix
#' @export
#'
#' @examples
snp_calculateProjections = function(PathOfNeuron, Hemisphere,
                                    StructureIDOfTargets, Annotation, Resolution,
                                    NumOfCores,
                                    OptionMetric = "LengthOfAxon") {
  library(nat)
  library(doSNOW)
  #library(doMC)
  library(foreach)
  
  NumNeuron = length(PathOfNeuron)
  BasicStructureID = StructureIDOfTargets
  NumOfTargets = length(StructureIDOfTargets)
  
  print(paste0("Number of neurons: ", NumNeuron))
  print(paste0("Number of regions: ", NumOfTargets))
  
  # order of axis is important
  NumRow_Anno = dim(Annotation)[1] # X
  NumColumn_Anno = dim(Annotation)[2] # Y
  NumSlice_Anno = dim(Annotation)[3] # Z
  
  
  cl = makeCluster(NumOfCores)
  registerDoSNOW(cl)
  #registerDoMC(cores = NumOfCores)
  
  pb = txtProgressBar(min = 0, max = NumNeuron, style = 3)
  progress = function(n) setTxtProgressBar(pb, n)
  opts = list(progress = progress)
  
  result = foreach(iNeuron = 1:NumNeuron, .combine=rbind, .inorder=TRUE, .options.snow = opts) %dopar% {
  #result = foreach(iNeuron = 1:NumNeuron, .combine=rbind, .inorder=TRUE) %dopar% {
    
    vector_proj = matrix(data = 0, nrow = 1, ncol = NumOfTargets)
    
    tmpSWC = read.table(PathOfNeuron[iNeuron], sep = " ",
                        col.names = c("NodeIndex","Label","X","Y","Z","W","Parent"))
    
    if(Hemisphere[iNeuron] == "Right")
      tmpSWC$Z = 11400 - tmpSWC$Z
    
    # support metric:
    if("LengthOfAxon" == OptionMetric) {
      numNode = nrow(tmpSWC)
      for(iNode in 1:numNode) {
        coord_node_current = c(tmpSWC$X[iNode], tmpSWC$Y[iNode], tmpSWC$Z[iNode])
        
        if(-1 == tmpSWC$Parent[iNode]) {
          coord_node_parent = coord_node_current
        } else {
          index = which(tmpSWC$NodeIndex == tmpSWC$Parent[iNode])
          coord_node_parent = c(tmpSWC$X[index], tmpSWC$Y[index], tmpSWC$Z[index])
        }
        
        
        ## what if the node is out of the annotation.
        tmpRow_current = floor(coord_node_current[1]/Resolution) + 1
        tmpColumn_current = floor(coord_node_current[2]/Resolution) + 1
        tmpSlice_current = floor(coord_node_current[3]/Resolution) + 1
        
        if(tmpRow_current > NumRow_Anno || tmpColumn_current > NumColumn_Anno || tmpSlice_current > NumSlice_Anno) {
          next
        }
        if(tmpRow_current < 1 || tmpColumn_current < 1 || tmpSlice_current < 1) {
          next
        }
        
        
        tmpRow_parent = floor(coord_node_parent[1]/Resolution) + 1
        tmpColumn_parent = floor(coord_node_parent[2]/Resolution) + 1
        tmpSlice_parent = floor(coord_node_parent[3]/Resolution) + 1
        
        if(tmpRow_parent > NumRow_Anno || tmpColumn_parent > NumColumn_Anno || tmpSlice_parent > NumSlice_Anno) {
          next
        }
        if(tmpRow_parent < 1 || tmpColumn_parent < 1 || tmpSlice_parent < 1) {
          next
        }
        
        structureID_node_current = Annotation[tmpRow_current, tmpColumn_current, tmpSlice_current]
        structureID_node_parent = Annotation[tmpRow_parent, tmpColumn_parent, tmpSlice_parent]
        
        if(FALSE == structureID_node_current %in% BasicStructureID ||
           FALSE == structureID_node_parent %in% BasicStructureID)
          next
        
        # whether the segment is in the same basic structure
        # todo what if it travels across two basic structures
        if(structureID_node_current == structureID_node_parent) {
          vector_proj[1, which(BasicStructureID == structureID_node_parent)] =
            vector_proj[1, which(BasicStructureID == structureID_node_parent)] +
            sqrt(sum((coord_node_current - coord_node_parent)^2))
        }
      }
    } # end of "LengthOfMetric" metric
    
    if("NumberOfEndNodes" == OptionMetric) {
      neuron = nat::read.neuron(PathOfNeuron[iNeuron])
      index_end_points = neuron$EndPoints
      
      for(iEnd in 1:length(index_end_points)) {
        coord_node_current = c(neuron$d$X[index_end_points[iEnd]],
                               neuron$d$Y[index_end_points[iEnd]],
                               neuron$d$Z[index_end_points[iEnd]])
        
        ## what if the node is out of the annotation.
        tmpRow_current = floor(coord_node_current[1]/Resolution) + 1
        tmpColumn_current = floor(coord_node_current[2]/Resolution) + 1
        tmpSlice_current = floor(coord_node_current[3]/Resolution) + 1
        
        if(tmpRow_current > NumRow_Anno || tmpColumn_current > NumColumn_Anno || tmpSlice_current > NumSlice_Anno) {
          next
        }
        
        structureID_node_current = Annotation[tmpRow_current, tmpColumn_current, tmpSlice_current]
        
        vector_proj[1, which(BasicStructureID == structureID_node_parent)] =
          vector_proj[1, which(BasicStructureID == structureID_node_parent)] + 1
      }
    } # end of "NumberOfEndNodes" metric
    
    
    if("NumberOfBranchNodes" == OptionMetric) {
      neuron = nat::read.neuron(PathOfNeuron[iNeuron])
      index_branch_points = neuron$BranchPoints
      
      for(iEnd in 1:length(index_branch_points)) {
        coord_node_current = c(neuron$d$X[index_branch_points[iEnd]],
                               neuron$d$Y[index_branch_points[iEnd]],
                               neuron$d$Z[index_branch_points[iEnd]])
        
        ## what if the node is out of the annotation.
        tmpRow_current = floor(coord_node_current[1]/Resolution) + 1
        tmpColumn_current = floor(coord_node_current[2]/Resolution) + 1
        tmpSlice_current = floor(coord_node_current[3]/Resolution) + 1
        
        if(tmpRow_current > NumRow_Anno || tmpColumn_current > NumColumn_Anno || tmpSlice_current > NumSlice_Anno) {
          next
        }
        
        structureID_node_current = Annotation[tmpRow_current, tmpColumn_current, tmpSlice_current]
        
        vector_proj[1, which(BasicStructureID == structureID_node_parent)] =
          vector_proj[1, which(BasicStructureID == structureID_node_parent)] + 1
      }
    } # end of "NumberOfBranchNodes" metric
    
    
    #
    # if(iNeuron %% 50 == 0)
    #   print(paste0(iNeuron, "/", NumNeuron))
    #
    # if (1) cat(paste0(iNeuron, "/", NumNeuron, "\n"))
    # flush.console()
    
    vector_proj
  }
  
  # close(pb)
  # stopCluster(cl)
  
  
  colnames(result) = BasicStructureID
  return(result)
}




time_start = Sys.time()

matrix_projection = snp_calculateProjections(PathOfNeuron = NeuronInfo$SWCPath, 
                                                  Hemisphere = NeuronInfo$Hemisphere,
                                                  StructureIDOfTargets = 1:NumCube,
                                                  Annotation = anno, 
                                                  Resolution = 100, 
                                                  NumOfCores = 8) 

rownames(matrix_projection) = NeuronInfo$Name
colnames(matrix_projection) = IndexCube

time_end = Sys.time()

print(time_end - time_start)
# write.csv(matrix_projection, "/home/gao/hy/pag_subdomain/output/matrix_projection.csv")
# save.image("/home/gao/hy/pag_subdomain/workfield.RData")
```
