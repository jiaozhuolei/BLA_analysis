```python
color_dict = {"Vglut1":'#107010',"Sst":'#0000ff',"Crh":'red'}
```

```python
# Extended Data Fig7a
for name_use in [["PL","ORB","ACA","ILA","FRP"]]:
        dfplot = df.loc[(df.name_use.isin(name_use))]
        dfplot2 = dfplot.pivot_table(index="filename",values=["soma_AP","soma_DV","soma_ML","AP","DV","ML"],aggfunc="mean")
        dfplot2["color"] = [color_dict[item.split("_")[2]] for item in dfplot2.index.to_list()]
        fig, axs = plt.subplots(3, 3, figsize=(15, 12))
        for i,soma_label in enumerate(["soma_AP","soma_DV","soma_ML"]):
                for j,ter_label in enumerate(["AP","DV","ML"]):
                        x = dfplot2[soma_label]
                        y = dfplot2[ter_label]

                        regression_result = linregress(x, y)
                        axs[i, j].scatter(x=x, y=y, s=10,c= dfplot2["color"])
                        axs[i, j].plot([min(x), max(x)],
                                [min(x) * regression_result.slope + regression_result.intercept,
                                max(x) * regression_result.slope + regression_result.intercept],
                                c='black',
                                linewidth=1.5,
                                dashes=[4, 2]
                                )
                        textstr = '\n'.join(['r = {:.3f}'.format(regression_result.rvalue),
                                'p = {:.3f}'.format(regression_result.pvalue),
                                "n = %d"%len(dfplot2)])
                        axs[i, j].text(x = 0.02,
                                y = 0.9,   
                                s = textstr,
                                transform=axs[i, j].transAxes, 
                                fontsize=12,
                                verticalalignment='top'
                                )

                        if soma_label.__contains__("ML"):
                                axs[i, j].invert_xaxis()

                        if ter_label.__contains__("ML"):
                                axs[i, j].invert_yaxis()


                        axs[i, j].set_title(soma_label+"-"+ter_label)
                        axs[i, j].set_xlabel(soma_label)
                        axs[i, j].set_ylabel(ter_label)
        name_uses = "PFC2"
        plt.suptitle('BLA_%s_all'%name_uses, fontsize=16, fontweight='bold')
        plt.subplots_adjust(wspace=0.25, hspace=0.35) 
        plt.savefig("BLA_%s_all.jpg"%name_uses,dpi=600)
        plt.savefig("BLA_%s_all.pdf"%name_uses,dpi=600)
        dfplot.to_csv("BLA_all_plot1_%s.csv"%name_uses)
        dfplot2.to_csv("BLA_all_plot2_%s.csv"%name_uses)
        plt.close()
                
```
