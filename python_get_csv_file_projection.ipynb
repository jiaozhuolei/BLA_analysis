```python
import os,nrrd,glob
import pandas as pd
import numpy as np
from math import floor,ceil
import threading
import pickle

```

```python
os.chdir(r"E:\0317 PAG")
readdata,header = nrrd.read("structure_795-100.nrrd",index_order='F')
# readdata[:,:,29:57] = 0
```

```python
IndexCube = np.where(readdata==1)
NumCube = len(IndexCube[0])
anno_PAG = np.zeros(readdata.shape, dtype = int)
for i in range(NumCube):
    anno_PAG[IndexCube[0][i],IndexCube[1][i],IndexCube[2][i]] = i+1
```

```python
# V3 add thread and the pickle

def get_cube_length(file,save_path,scale=10*10):
    # print(file)
    # print(process)
    filename = file.split("\\")[-1]
    dfswc = pd.read_csv(file,sep=" ",header=None)
    dfswc.columns = ["ID","type","AP","DV","ML","R","parentID"]
    dictmp = {}
    for i in range(NumCube):
        dictmp[i+1] = 0   
    for i in dfswc.ID.tolist():
        # index_PAG = readdata[int(floor(dfswc.loc[dfswc.ID ==i].AP.values[0]/scale)),int(floor(dfswc.loc[dfswc.ID ==i].DV.values[0]/scale)),
        #             int(floor(dfswc.loc[dfswc.ID ==i].ML.values[0]/scale))]
        # if index_PAG == 1:
        try:
            Indextmp = anno_PAG[int(floor(dfswc.loc[dfswc.ID ==i].AP.values[0]/scale)),int(floor(dfswc.loc[dfswc.ID ==i].DV.values[0]/scale)),
                    int(floor(dfswc.loc[dfswc.ID ==i].ML.values[0]/scale))]
            if i == 1:
                parent = 1
            else:
                parent = dfswc.loc[dfswc.ID ==i].parentID.values[0]
            if Indextmp != 0:
                dictmp[Indextmp] += np.sqrt(np.square(dfswc.loc[dfswc.ID ==i].AP.values[0]-dfswc.loc[dfswc.ID ==parent].AP.values[0]) + 
                        np.square(dfswc.loc[dfswc.ID ==i].DV.values[0]-dfswc.loc[dfswc.ID ==parent].DV.values[0]) +
                        np.square(dfswc.loc[dfswc.ID ==i].ML.values[0]-dfswc.loc[dfswc.ID ==parent].ML.values[0]))
        except Exception as e:
            print (e)
    sum_dist = 0
 
    for a in dictmp:
        sum_dist += dictmp[a]
    if sum_dist >100:
        PklPath = os.path.join(save_path,filename.split(".")[0]+".pkl")
        with open(PklPath,"wb") as f:
            pickle.dump(dictmp,f)
   

scale = 10*10

filepath = r"E:\all neurons 0312\rawdata\HY_new\*.swc"
swcfiles = glob.glob(filepath)
num_file = len(swcfiles)
save_path = r"E:\0317 PAG\all_pkl"
threads = []
step = 50

num_pro = ceil(num_file/step)

for j in range(num_pro):
    if j < num_pro-1:
        for n,file in enumerate(swcfiles[j*step:(j+1)*step]):
            t = threading.Thread(target=get_cube_length, args=(file,save_path))
            t.start()
            threads.append(t)
        print(j*step)
        for thread in threads:
            thread.join()
    else:
        for n,file in enumerate(swcfiles[j*step:]):
            t = threading.Thread(target=get_cube_length, args=(file,save_path))
            t.start()
            threads.append(t)
        print(j*step)
        for thread in threads:
            thread.join()


```
