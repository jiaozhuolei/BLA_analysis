```python
# all for PAG
files = glob.glob(r"H:\20240822_figs\20240827_PVH\csv_new\*.csv")
for file in tqdm(files):
    filename = file.split("\\")[-1].split(".")[0]
    dfn = pd.read_csv(file,index_col=0)
    # dft = dfn.loc[(dfn["terminal"]==1)&(dfn["ML"]<=5700)&((dfn["region"]=="AM")|(dfn["region"]=="AD")|(dfn["region"]=="AV")|(dfn["region"]=="VTN"))]
    dft = dfn.loc[(dfn["terminal"]==1)&((dfn["name_use"].isin(["ME"])))]
    try:
        last = dft.ID.tolist()[-1]
        IDlast = []
        IDlast.append(last)

        while last != 1:
            last = dfn.loc[dfn.ID==last].parent.values[0]
            IDlast.append(last)

        IDs = IDlast.copy()
        for IDt in dft.ID.tolist()[0:-1]:
            while IDt not in IDs:
                IDs.append(IDt)
                IDt = dfn.loc[dfn.ID==IDt].parent.values[0]

        dff = dfn.loc[dfn.ID.isin(IDs)]
        dff.to_csv(r"H:\20240822_figs\20240827_PVH\ME_csv\%s_ME.csv"%filename)
        
    except:
        print(len(dft)) 
        print(file)     
```
