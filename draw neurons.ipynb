```python
import plotly.graph_objects as go
import pandas as pd
import sys,os,glob,shutil
import numpy as np
import random
import matplotlib.pyplot as plt
sys.path.append(r"D:\python\core")
import visualization as vs
from tqdm import tqdm
```

```python
dfdata = pd.read_excel("ACA_projecting_neurons_chose.xlsx",index_col=0)
```

```python
area = "ACA"
# dff = dfdata.loc[dfdata.location == area]
dff = dfdata
neuron_name_list = dff.index.to_list()
try:
    os.mkdir(".\\%s"%area)
except:
    pass
savepath = ".\\%s"%area
```

```python
neuron_name_list
```

```python
cluster = 5
# plot_areas = ["PVT","MOB"]
plot_areas = ["PAG","LC"]
region_add = False
try:
    os.mkdir(".\\figuse\\cluster_%s"%cluster)
except:
    pass
savepath = ".\\figuse\\cluster_%s"%cluster
plot_name = ""
for i in range(len(plot_areas)):
    plot_name = plot_name+plot_areas[i]+"_"
dfc = dfdata.loc[dfdata.cluster == cluster][plot_areas]
dfc.loc[:,"sum"] =dfc.apply(lambda x:x.sum(),axis =1)
dff = dfc.sort_values(by = "sum",axis=0,ascending=False)
neuron_name_list = dff.index.to_list()[0:7]
```

```python
# add the classfication results (neuoron cluster)
# savepath = ".\\figneuron"
dfcsvpath = ".\\csv"
# plot_areas = ["MY","unname"]
# plot_areas = ["CA1","AHN"]
plot_areas = ["ACA"]
plot_colors = ["gold","red","cyan","gold","red","yellowgreen","violet"]
region_add = False


for neuron_name in tqdm(neuron_name_list):
    neuron_name = neuron_name.split(".")[0]
    file = os.path.join(dfcsvpath,neuron_name+".csv")
    vols = []
    if region_add == True:
        for vol_area,vol_color in zip(plot_areas,plot_colors):
            vols.append(vs.generate_go_ba(area = vol_area,o = 0.2,c="gray"))        
    vols.append(vs.generate_go_ba(area = "root",o = 0.2,c="lightgray"))
    # vols.append(vs.generate_go_ba(area = "root",o = 0.2,c="gray"))
    dftmp = pd.read_csv(file,sep = ",",index_col=0)
    # dftmp = dfneuron.loc[dfneuron.in_not ==1]
    # c = vs.generate_color()
    for edge in set(dftmp.bratch_ID.tolist()):
        dfnew = dftmp.loc[dftmp.bratch_ID == edge]
        vols.append(vs.generate_go_lines(dfnew[["AP","DV","ML"]],c="black",name=""))

    for plot_area,plot_color in zip(plot_areas,plot_colors):
        dftmp_area = dftmp.loc[dftmp.name_use == plot_area]
        for i,edge_area in enumerate(set(dftmp_area.bratch_ID.tolist())):
            if i == 0:
                dfnew_area = dftmp_area.loc[dftmp_area.bratch_ID == edge_area]
                vols.append(vs.generate_go_lines(dfnew_area[["AP","DV","ML"]],c=plot_color,name=plot_area,showlegend=True))
            else:
                dfnew_area = dftmp_area.loc[dftmp_area.bratch_ID == edge_area]
                vols.append(vs.generate_go_lines(dfnew_area[["AP","DV","ML"]],c=plot_color,name=plot_area,showlegend=False))

    soma = dftmp.loc[dftmp.type ==1][["AP","DV","ML"]]
    vols.append(vs.generate_go_scatters(soma,color="red",name=""))
    fig = go.Figure(vols)
    for eyes in ["front","left","top","se"]:
        # for eyes in ["front","left"]:
        layout = vs.generate_go_layout(title="",eye_from=eyes)
        fig.layout = layout
        fig.write_image(os.path.join(savepath,"%s_%s.jpg"%(neuron_name,eyes)),scale =8)  
```
