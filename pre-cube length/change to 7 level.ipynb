```python
# search_brain_area返回整个area的list
areas_dic = []
for area in dfchange.region:
    level = {}
    level["father"] = ba.search_brain_area(area)[7][1]
    level["child"] = area
    areas_dic.append(level)
dfmap = pd.DataFrame(areas_dic)
```

```python
dfcluster = pd.read_excel(r"E:\20220210_for_figs\ploty soma plot\cluster_all_30_more.xlsx",index_col=0)
```

```python
dfcluster["genotype"] = dfcluster["genotype"].str.capitalize().str.replace("-cre","")
```

```python
def nor_level_7(x):
    # print(x["location"])
    if x["location"] in dfmap.child.tolist():
        return dfmap.loc[dfmap.child == x["location"]].father.tolist()[0]
    if x["location"] not in dfmap.child.tolist():
        return x["location"]

# 注意需要设置axis==1，这是series的index是columns
dfcluster.loc[:,"location2"] = dfcluster.apply(nor_level_7, axis=1)
```
