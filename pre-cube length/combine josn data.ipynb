```python
# module 5 提取soma location
import os,json,glob
import pandas as pd
```

```python
jsonpath = r"E:\20220210_for_figs\long-range-new\json_for_long_range\*"
savepath = r"E:\20220210_for_figs\long-range-new"
# dfproject = pd.DataFrame()
jsondata = glob.glob(jsonpath)
col = []
for file in jsondata:
    with open(file, 'r') as rf:
        data = json.load(rf)
        soma = {}
        filename = file.split("\\")[-1].split(".")[0]
        soma["file"] = filename
        soma["neurons"] = filename[0:10]
        soma["location"] = filename.split("_")[-1]
        soma["geno"] = filename.split("_")[-2]
        soma["AP"] = data["somapoint"][0]
        soma["DV"] = data["somapoint"][1]
        if data["somapoint"][2] >5700:
            soma["ML"] = 11400-data["somapoint"][2]
        else:
            soma["ML"] = data["somapoint"][2]
        col.append(soma)
dfsoma = pd.DataFrame(col)
dfsoma = dfsoma.set_index("neurons")
dfsoma.to_excel(os.path.join(savepath,"long_range_soma_infor.xlsx"))
```

```python
# json get soma areas
jsonpath = r"E:\all_neurons_0616\json\*"
savepath = r"E:\all_neurons_0616"
# dfproject = pd.DataFrame()
jsondata = glob.glob(jsonpath)
col = []
for file in jsondata:
    with open(file, 'r') as rf:
        data = json.load(rf)
        soma = {}
        filename = file.split("\\")[-1].split(".")[0]
        soma["file"] = filename
        soma["neurons"] = filename[0:10]
        soma["location"] = filename.split("_")[-1]
        soma["somaregion"] = data["somaregion"]
        soma["geno"] = filename.split("_")[-2]
        soma["AP"] = data["somapoint"][0]
        soma["DV"] = data["somapoint"][1]
        if data["somapoint"][2] >5700:
            soma["ML"] = 11400-data["somapoint"][2]
        else:
            soma["ML"] = data["somapoint"][2]
        col.append(soma)
dfsoma = pd.DataFrame(col)
dfsoma = dfsoma.set_index("neurons")
dfsoma.to_excel(os.path.join(savepath,"all_soma_infor.xlsx"))
```

```python
# 用坐标去find soma location
import nrrd,math
readdata,header = nrrd.read(r"D:\projectome\annotation_10_2017.nrrd")
dfarea = pd.read_excel(r"D:\python\pre-cube length\brain_areas.xlsx")

```

```python
def get_wendu_type(x):
    # count = readdata[math.ceil(x["AP"]/10),math.ceil(x["DV"]/10),math.ceil(x["ML"]/10)]
    count = readdata[math.ceil(x["AP"]/10),math.ceil(x["DV"]/10),math.ceil(x["ML"]/10)]
    area = dfarea.loc[dfarea.ID == count].region.tolist()[0]
    return area

# 注意需要设置axis==1，这是series的index是columns
dfsoma.loc[:, "ceil_soma"] = dfsoma.apply(get_wendu_type, axis=1)
```
